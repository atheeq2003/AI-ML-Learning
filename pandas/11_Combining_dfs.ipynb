```python
# Combining dataframes - concat(), merge(), join()
```

```python
datasets = {
    "Southern_Stations: " : southern_stations,
    "Western_Stations: " : western_stations,
    "Train Routes:" : train_routes,
    "Train Schedule:" : train_schedule,
    "Station Facilities:" : station_facilities
}

for name, df in datasets.items():
    print(name, df.shape)
```

```python
# concat() - combine 2 or more df into single df
# Vertically - Based on Rows
# Horizontally - Based on Columns
```

```python
south_sample = southern_stations[["Station_ID","Station_Name"]]
west_sample = western_stations[["Station_ID", "Station_Name"]]

horizontal_combine = pd.concat([south_sample, west_sample], axis = 1)
horizontal_combine.head()
```

```python
south_sample = southern_stations[["Station_ID","Station_Name"]]
west_sample = western_stations[["Station_ID", "Station_Name"]]

vertical_combine = pd.concat([south_sample, west_sample], axis = 0)
vertical_combine.head()
```
