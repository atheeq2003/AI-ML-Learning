# *Task 1 – 1D Array: Daily Water Consumption Analysis*
---

```python
import numpy as np

water_consumption = [1800, 2200, 1950, 2500, 2100, 1700, 2300, 2000, 2400, 2150]
```

```python
# 1. Convert the given water-consumption data into a NumPy array and display the complete array.

array = np.array(water_consumption)
array
```

```python
# 2. Find the number of dimensions, number of elements, shape, and data type of the array.

print(array.ndim)
print(array.size)
print(array.shape)
print(array.dtype)
```

```python
# 3. Display the water consumption recorded on the 4th day.

print(array)
print(array[3])
```

```python
# 4. Display the water consumption recorded on the last day, using indexing from the end.

print(array)
print(array[-1])
```

```python
# 5. Display the water consumption recorded from Day 3 to Day 7.

print(array)
print(array[2:7])
```

```python
# 6. Display the consumption recorded during the first 5 days.

print(array)
print(array[:5])
```

```python
# 7. Display the consumption recorded during the last 3 days.

print(array)
print(array[-3:])
```

```python
# 8. Create a NumPy array containing the day numbers from 1 to 10.

days = np.arange(1, 11)
print(days)
```

```python
# 9. Create a NumPy array containing 5 zero values to represent days for which water consumption has not yet been recorded.

unrecorded_days = np.zeros(5)
print(unrecorded_days)
```

```python
# 10. Create a NumPy array containing 5 one values to represent completed daily records.

completed_daily_records = np.ones(5)
print(completed_daily_records)
```

```python
# 11. Create a NumPy array containing 5 values, where every value is 2000, representing the recommended daily water-consumption target.

daily_target = np.full(5, 2000)
print(daily_target)
```

```python
# 12. Create an empty NumPy array with 5 elements for temporary data storage.

temp_storage = np.empty(5)
print(temp_storage)
```

```python
# 13. Calculate the total amount of water consumed during the 10 days.

total_amount = np.sum(array)
print(total_amount)
```

```python
# 14. Calculate the average and median daily water consumption.

average = np.mean(array)
median = np.median(array)

print(average)
print(median)
```

```python
# 15. Find the minimum, maximum, and variance of the daily water-consumption values.

minimum = np.min(array)
maximum = np.max(array)
variance = np.var(array)

print(minimum)
print(maximum)
print(variance)
```
