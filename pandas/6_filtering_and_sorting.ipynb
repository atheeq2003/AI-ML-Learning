# *Part A: Filtering Operations*
---

```python
# 1. Single Condition Filtering
```

```python
# 1. Filter employees whose Efficiency_Percentage is greater than 80.

df[df["Efficiency_Percentage"] > 80]
```

```python
# 2. Filter employees working in the Production department.

df[df["Department"] == "Production"]
```

```python
# 3. Filter employees located in Chennai.

df[df["Location"] == "Chennai"]
```

```python
# 4. Filter employees with more than 5 years of experience.

df[df["Experience_Years"] > 5]
```

```python
# 5. Filter employees whose Work_Status is Pending.

df[df["Work_Status"] == "Pending"]
```

```python
# 2. Multiple Conditions
```

```python
# 6. Find employees from the Production department and working in the Morning shift.

df[(df["Department"] == "Production") & (df["Shift"] == "Morning")]
```

```python
# 7. Find employees with High Priority and whose work status is In Progress.

df[(df["Priority"] == "High") & (df["Work_Status"] == "In Progress")]
```

```python
# 8. Find employees who have more than 8 years of experience and efficiency greater than 70%.

df[(df["Experience_Years"] > 8) & (df["Efficiency_Percentage"] > 70)]
```

```python
# 9. Filter employees located in Chennai or Coimbatore.

df[(df["Location"] == "Chennai") | (df["Location"] == "Coimbatore")]
```

```python
# 10. Filter employees working in the Morning or Night shift.

df[(df["Shift"] == "Morning") | (df["Shift"] == "Night")]
```

```python
# 11. Find employees whose Work_Status is Pending or In Progress.

df[(df["Work_Status"] == "Pending") | (df["Work_Status"] == "In Progress")]
```

```python
# 12. Filter employees belonging to these departments: Production, Maintenance, Quality Control

df[df["Department"].isin(["Production", "Maintenance", "Quality Control"])]
```

```python
# 13. Filter employees located in: Chennai, Bengaluru, Hyderabad

df[df["Location"].isin(["Chennai", "Bengaluru", "Hyderabad"])]
```

```python
# 14. Find employees whose experience is between 3 and 7 years.

df[df["Experience_Years"].between(3, 7)]
```

```python
# 15. Find employees whose efficiency percentage is between 60 and 80.

df[df["Efficiency_Percentage"].between(60, 80)]
```

```python
# 16. Filter employees whose names start with "A".

df[df["Employee_Name"].str.startswith("A")]
```

```python
# 17. Filter employees whose names contain "Priya".

df[df["Employee_Name"].str.contains("Priya")]
```

```python
# 18. Use the query() method to find employees with: Efficiency_Percentage greater than 75, Experience_Years greater than 5

df.query("Efficiency_Percentage > 75 and Experience_Years > 5")
```

```python
# 19. Use query() to filter employees from Chennai with High priority.

df.query("Priority == 'High' and Location == 'Chennai'")
```
