```python
import pandas as pd

data = {
    "patient_name": ["Arun", "Priya", "Rahul", "Divya", "Karthik"],
    "age": [25, 26, 56, 39, 60],
    "wait_time": [15, 45, 30, 60, 10],
    "consultation_minutes": [20, 35, 25, 40, 15]
}

df = pd.DataFrame(data)
df.head()
```

```python
# Applying a function to single column
# Increase waiting time

def add_five_mins(wait_time):
    return wait_time + 5

df["updated_wait_time"] = df["wait_time"].apply(add_five_mins)
df
```

```python
# Using lambda function with apply()

df["updated_plus"] = df["wait_time"].apply(lambda x: x+10)
df
```

```python
import pandas as pd

employee = pd.DataFrame({
    "employee_id": [101, 102, 103, 104, 105],
    "employee_name": ["Arun", "Divya", "Raghul", "Karthik", "Varun"],
    "dept": ["IT", "FN", "SL", "HR", "IT"]
})

print(employee)
```

```python
# dept_map = {
#     "IT" : "Information Technology"
# }

employee["dept_name"] = employee["dept"].map({
    "IT" : "Information Technology",
    "HR" : "Human Resources",
    "SL" : "Sales",
    "FN" : "Finance"
})

employee
```

```python
employee["code_length"] = employee["dept"].map(lambda x: len(x))
employee
```

```python
df["age_group"] = df["age"].map(lambda x: "Young" if x < 50 else "Adult")
df
```
