```python
import pandas as pd

df = pd.read_csv(r"library_book.csv")
df.head()
```

```python
# groupby() - Divide the rows into groups to perform some calculations

df.shape
```

```python
df.info()
```

```python
df.groupby("Department")["Fine_Amount"].sum()
```

```python
avg_late_days = df.groupby("Department")["Days_Late"].sum()
avg_late_days
```

```python
transaction_count = df.groupby("Department")["Transaction_ID"].count()
transaction_count
```

```python
books = df.groupby("Book_Category")["Book_Price"].agg(["max", "min", "mean"])
books
```

```python
computer_science = df.groupby(["Department", "Return_Status"])["Transaction_ID"].count()
computer_science
```

```python
total_fine = df.groupby(["Department", "Return_Status"])["Fine_Amount"].sum()
total_fine
```

```python
summary_of_transactions = df.groupby("Book_Category").agg(
    transaction = ("Transaction_ID", "count"),
    total_fine = ("Fine_Amount", "sum"),
    avg_late_days = ("Days_Late", "mean"),
    max_late_days = ("Days_Late", "max")
)

summary_of_transactions
```
