```python
# Importing pandas lib and loding csv
import pandas as pd

raw_data = pd.read_csv(r"university_event_registration_lab 8.csv")

raw_data.head()
```

```python
# Copying

raw_df = raw_data.copy()
clean_df = raw_df.copy()
```

```python
clean_df.columns
```

```python
# renaming columns

clean_df = clean_df.rename(columns = {
    'Student ID' :'student_id', 
    'Student Name':'student_name', 
    'Gender': 'gender',
    'Event Date' : 'event_date', 
    'Age': 'age', 
    'City': 'city',
    'Department': 'department', 
    'Event Type': 'event_type', 
    'Registration Date': 'registration_date', 
    'Registration Channel':'registration_channel',
    'Payment Status': 'payment_status', 
    'Attendance Status':'attendance_status', 
    'Meal Preference':'meal_preference',
    'Registration Fee':'registration_fee', 
    'Event Duration Hours':'event_duration_hours', 
    'Feedback Rating':'feedback_rating'
})
print(clean_df.columns)
```

```python
# Finding Null Values column wise

clean_df.isnull().sum()
```

```python
# Finding null values row wise

clean_df.isnull().any(axis=1).sum()
```

```python
clean_df.info()
```

```python
# Filling Null Values

clean_df['gender'] = clean_df['gender'].fillna("Unknown")
clean_df['city'] = clean_df['city'].fillna("Unknown")
clean_df["department"] = clean_df["department"].fillna("Unknown")
clean_df["registration_channel"] = clean_df["registration_channel"].fillna("Unknown")
clean_df["meal_preference"] = clean_df["meal_preference"].fillna("Unknown")

clean_df.isnull().sum()
```

```python
# Converting String to numeric dtype

clean_df['feedback_rating'] = pd.to_numeric(clean_df['feedback_rating'], errors = 'coerce')
clean_df['event_date'] = pd.to_datetime(clean_df['event_date'], format = 'mixed', errors = 'coerce')
clean_df["registration_date"] = pd.to_datetime(clean_df["registration_date"],format = 'mixed', errors = 'coerce')
clean_df['event_duration_hours'] = pd.to_numeric(clean_df['event_duration_hours'],errors ='coerce')
clean_df['registration_fee'] = pd.to_numeric(clean_df['registration_fee'],errors ='coerce')

clean_df.info()
```

```python
clean_df.head()
```

```python
print(clean_df.isna().sum())
```

```python
# Remove Extra Spaces

for columns in clean_df.select_dtypes(include='str'):
    clean_df[columns] = clean_df[columns].str.lower().str.strip()

clean_df.head()
```

```python
clean_df['event_duration_hours'].dtypes
```

```python
# Filling date columns

clean_df["event_date"] = clean_df["event_date"].fillna(pd.Timestamp("1900-01-01"))
clean_df["registration_date"] = clean_df["registration_date"].fillna(pd.Timestamp("1900-01-01"))
clean_df.isnull().sum()
```

```python
# Filling missing values using mean and median

event_duration_hrs_median = clean_df['event_duration_hours'].median()
clean_df['event_duration_hours'] = clean_df['event_duration_hours'].fillna(event_duration_hrs_median)

feedback_rating_mean = clean_df['feedback_rating'].mean()
clean_df['feedback_rating'] = clean_df['feedback_rating'].fillna(feedback_rating_mean.round())

clean_df.isnull().sum()
```

```python
# Checking Unique Values

print("Gender \n", clean_df["gender"].unique())
print("Age \n", clean_df["age"].unique())
print("Cities \n", clean_df["city"].unique())
print("Department \n", clean_df["department"].unique())
print("Event Type \n", clean_df["event_type"].unique())
print("Reg Channel \n", clean_df["registration_channel"].unique())
print("Payment Status \n", clean_df["payment_status"].unique())
print("Attendance Status \n", clean_df["attendance_status"].unique())
print("Meal Pref \n", clean_df["meal_preference"].unique())
print("Registration Fee \n", clean_df["registration_fee"].unique())
print("Event Duration Hours \n", clean_df["event_duration_hours"].unique())
print("Feedback Rating \n", clean_df["feedback_rating"].unique())
```

```python
# Replacing inconsistent values and filling with mean, median and unknown. Round of feedback rating and age

clean_df['city'] = clean_df['city'].replace(['','nan'], pd.NA)
clean_df["city"] = clean_df["city"].fillna("unknown")
clean_df["department"] = clean_df["department"].replace('nan', pd.NA)
clean_df["department"] = clean_df["department"].fillna("unknown")
clean_df["registration_channel"] = clean_df["registration_channel"].replace('nan', pd.NA)
clean_df["registration_channel"] = clean_df["registration_channel"].fillna("unknown")
clean_df["payment_status"] = clean_df["payment_status"].replace('waive', 'waived')
clean_df['meal_preference'] = clean_df['meal_preference'].replace(['','nan'], pd.NA)
clean_df['meal_preference'] = clean_df['meal_preference'].replace('no meal', "no-meal")
clean_df["meal_preference"] = clean_df["meal_preference"].fillna("unknown")
clean_df["registration_fee"] = clean_df["registration_fee"].replace('nan', pd.NA)
clean_df["registration_fee"] = clean_df["registration_fee"].fillna(clean_df["registration_fee"].median())
clean_df["age"] = clean_df["age"].replace(-5, round(clean_df["age"].mean()))
```

```python
# Checking Unique Values

print("Gender \n", clean_df["gender"].unique())
print("Age \n", clean_df["age"].unique())
print("Cities \n", clean_df["city"].unique())
print("Department \n", clean_df["department"].unique())
print("Event Type \n", clean_df["event_type"].unique())
print("Reg Channel \n", clean_df["registration_channel"].unique())
print("Payment Status \n", clean_df["payment_status"].unique())
print("Attendance Status \n", clean_df["attendance_status"].unique())
print("Meal Pref \n", clean_df["meal_preference"].unique())
print("Registration Fee \n", clean_df["registration_fee"].unique())
print("Event Duration Hours \n", clean_df["event_duration_hours"].unique())
print("Feedback Rating \n", clean_df["feedback_rating"].unique())
```
