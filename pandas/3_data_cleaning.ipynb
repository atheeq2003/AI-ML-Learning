```python
# Handle Missing Values
# Method 1: dropna() - Remove rows - Not Mostly Applicable
# Method 2: fillna() - Fill with Constant values (Fixed Value)
# Method 3: Fill Numerical values with mean
# Method 4: Fill Numberical values with median
# Method 5: Outliers
```

```python
# Method 2: fillna() - Fill with Constant values (Fixed Value)

clean_df["Gender"] = clean_df["Gender"].fillna("Unknown")
```

```python
clean_df.isnull().sum()
```

```python
missing_genders = clean_df[
    clean_df["Gender"].isnull()
]
missing_genders.head(20)
```

```python
clean_df["City"] = clean_df["City"].fillna("Unknown")
clean_df["Doctor Name"] = clean_df["Doctor Name"].fillna("Not Assigned")
clean_df["Insurance Type"] = clean_df["Insurance Type"].fillna("Unknown")
clean_df["Booking Channel"] = clean_df["Booking Channel"].fillna("Not Available")
clean_df["Follow Up Required"] = clean_df["Follow Up Required"].fillna("Unknown")
clean_df["Symptoms"] = clean_df["Symptoms"].fillna("Not Found")
```

```python
clean_df.isnull().sum()
```

```python
# Convert the column to numerical

clean_df["Wait Time Minutes"] = pd.to_numeric(
    clean_df["Wait Time Minutes"],
    errors = "coerce"
)

clean_df["Wait Time Minutes"].dtype
```

```python
# Method 3: Fill Numerical values with mean

wait_time_mean = clean_df["Wait Time Minutes"].mean()
wait_time_mean
```

```python
# Method: Handling Missing Values using Forward Fill and Backward Fill
data = pd.DataFrame({
    "Employee":["Arun", "Priya", "Rahul", "Divya", "Venkat", "Karthik"],
    "Dept": ["IT", None, "HR", None, "Sales", "HR"],
    "Salary": [20000, 21000, None, 45000, None, None]
})

data
```

```python
data.isnull().sum()
```

```python
# Forward Fill - ffill()

data["Dept"] = data["Dept"].ffill()
data
```

```python
# Backward Fill - bfill()

data["Salary"] = data["Salary"].bfill()
data
```

```python
data["Salary"] = data["Salary"].ffill()
data
```

```python
data.isnull().sum()
```
