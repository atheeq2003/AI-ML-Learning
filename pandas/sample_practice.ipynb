```python
# Series - One dimensional Data

import pandas as pd

attendance = pd.Series([98,34,255,32], index = [1, 2, 3, 4])
print(attendance)
```

```python
# Dataframe - Two Dimensional Data

students = {
    "Student_ID" : ["ST101", "ST102", "ST103"],
    "Name" : ["Atheeq", "Abdur", "Hari"],
    "Course" : ["DL", "CN", "UI"]
}

print(students)

# Convert dict to data frame - DataFrame()
data = pd.DataFrame(students, index = [1, 2, 3])
data
```

```python
data = pd.DataFrame({
    "Student_ID" : ["ST101", "ST102", "ST103"],
    "Name" : ["Atheeq", "Abdur", "Hari"],
    "Course" : ["DL", "CN", "UI/UX"]
})
data
```
