```python
# Create arrays using NumPy array-generation functions
import numpy as np

print("Sensor IDs")
print(np.arange(101, 111))

print("\nMachine Status Matrix")
print(np.ones((3, 4)))

print("\nFault Matrix")
print(np.zeros((3, 4)))

print("\nDefault Temperature")
print(np.full((3, 4), 25))

print("\nMemory Allocation")
print(np.empty((2, 3)))

print("\nCalibration Levels")
print(np.linspace(0, 100, 6))

print("\nIdentity Matrix")
print(np.eye(4))
```

```python
# Create and display NumPy matrices
import numpy as np

pixel_values = np.array([
    [120, 130, 140],
    [150, 160, 170],
    [180, 190, 200]
])

# Display the original matrix
print("Original Image Matrix")
print(pixel_values)

# Create a zero matrix
print("\nBlank Image")
print(np.zeros((3, 3)))

# Create a matrix filled with 255
print("\nWhite Image")
print(np.full((3, 3), 255))

# Create an identity matrix
print("\nIdentity Filter")
print(np.identity(3))
```

```python
# Create a 2D NumPy array and inspect its attributes
import numpy as np

patients = np.array([
    [98.6, 72, 120],
    [99.1, 76, 118],
    [100.2, 80, 122],
    [98.4, 70, 115]
])

# Display array attributes
print("Patient Data")
print(patients)
print("\nDimensions :", patients.ndim)
print("Shape :", patients.shape)
print("Total Elements :", patients.size)
print("Data Type :", patients.dtype)
print("Memory Per Element :", patients.itemsize, "Bytes")
print("Total Memory :", patients.nbytes, "Bytes")

# Transpose the array
print("\nTranspose")
print(patients.T)

# Convert the array to float32
print("\nConvert to Float32")
print(patients.astype(np.float32))
```

```python
# Create a NumPy array and access elements using indexing and slicing
import numpy as np

students = np.array([
    ["Aarav", 85, 90, 88],
    ["Diya", 78, 82, 80],
    ["Kabir", 92, 95, 94],
    ["Meera", 89, 91, 90]
], dtype=object)

# Display selected elements, rows, columns, and slices
print("Student Dataset\n")
print(students)
print("\nFirst Student")
print(students[0])
print("\nLast Student")
print(students[-1])
print("\nSecond Student Name")
print(students[1, 0])
print("\nMarks Column")
print(students[:, 1])
print("\nFirst Two Students")
print(students[0:2])
print("\nName and First Two Marks")
print(students[:, 0:3])
```

```python
# Create a NumPy array and select rows and columns
import numpy as np

sales = np.array([
    [101, 1200, 35],
    [102, 1800, 28],
    [103, 2400, 45],
    [104, 1500, 32]
])

# Display the array and selected data
print("Sales Dataset\n")
print(sales)
print("\nProduct IDs")
print(sales[:, 0])
print("\nPrices")
print(sales[:, 1])
print("\nUnits Sold")
print(sales[:, 2])
print("\nFirst Three Products")
print(sales[:3])
print("\nPrice and Units Columns")
print(sales[:, 1:3])
```

```python
# Calculate mathematical and statistical values from a NumPy array
import numpy as np

salary = np.array([42000, 48000, 51000, 45000, 60000, 55000])

# Display statistical calculations
print("Employee Salaries")
print(salary)
print("\nTotal Salary :", np.sum(salary))
print("Average Salary :", np.mean(salary))
print("Median Salary :", np.median(salary))
print("Highest Salary :", np.max(salary))
print("Lowest Salary :", np.min(salary))
print("Salary Standard Deviation :", np.std(salary))
print("Salary Variance :", np.var(salary))
print("Highest Salary Index :", np.argmax(salary))
print("Lowest Salary Index :", np.argmin(salary))
```

```python
# Create arrays using NumPy array-generation functions
import numpy as np

print("Parking Slot Numbers")
print(np.arange(1, 21))

print("\nEmpty Parking Slots")
print(np.zeros((4, 5)))

print("\nOccupied Parking Slots")
print(np.ones((4, 5)))

print("\nParking Charges")
print(np.full((4, 5), 50))

print("\nEmpty Matrix")
print(np.empty((3, 4)))

print("\nParking IDs")
print(np.arange(100, 201, 10))

print("\nEqually Spaced Values")
print(np.linspace(0, 1, 8))

print("\nIdentity Matrix using eye()")
print(np.eye(5))

print("\nIdentity Matrix using identity()")
print(np.identity(5))
```
