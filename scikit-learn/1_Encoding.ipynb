# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
products = ["Key", "Mouse", "Laptop", "Pendrive", "Laptop"]
encoder = LabelEncoder()
encoded_prods = encoder.fit_transform(products)
encoded_prods
```

# One Hot Encoding

```python
# from sklearn.preprocessing import OneHotEncoder
import pandas as pd

color = pd.DataFrame({
    "Color": ["Red", "Blue", "Yellow", "Red"]
})

encoded_colors = pd.get_dummies(color, columns=["Color"])
encoded_colors
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

color = pd.DataFrame({
    "Color": ["Red", "Blue", "Yellow", "Red"]
})
encoder = OneHotEncoder(sparse_output=False) # Setting to false will disable regular array output and gives sparse matrix output
encoded_colors = encoder.fit_transform(color[["Color"]])
encoded_colors
```

```python
color = pd.DataFrame({
    "Color": ["Red", "Blue", "Yellow", "Red"]
})
encoder = OneHotEncoder(sparse_output=False)
encoded_colors = encoder.fit_transform(color[["Color"]])
encoded_df = pd.DataFrame(
    encoded_colors,
    columns = encoder.get_feature_names_out(["Color"])
)

encoded_df
# [1001, 0100, 0010, 1001]
```

# Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
data = pd.DataFrame({
    "Qualifications": ["Bachelors", "School", "PHD", "Masters"]
})
encoder = OrdinalEncoder(categories=[["School", "Bachelors", "Masters", "PHD"]])

data["Ordered Qualifications"] = encoder.fit_transform(data[["Qualifications"]])
data
```
