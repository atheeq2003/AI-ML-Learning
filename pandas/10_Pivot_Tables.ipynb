```python
import pandas as pd
```

```python
df = pd.read_csv(r"Training_center.csv")
df.head()
```

```python
df.info()
```

```python
prgrm_by_student = df.pivot_table(
    columns="Program",
    values="Student_ID",
    aggfunc="count"
)

prgrm_by_student
```

```python
prgrm_by_student = prgrm_by_student.rename(
    columns={"Student_ID": "student_count"}
)
prgrm_by_student
```

```python
program_gender = df.pivot_table(
    index = "Program",
    columns = "Gender",
    values = "Student_ID",
    aggfunc = 'count'
)

program_gender
```
