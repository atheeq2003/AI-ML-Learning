```python
import pandas as pd
df = pd.read_csv(r"airline_booking.csv")
df.head()
```

---
# *Part A – GroupBy*
---

```python
# 1. Total Fare by Airline - Find the total Base Fare collected by each Airline using groupby().

total_base_fare_airline = df.groupby("Airline")["Base_Fare"].sum()
total_base_fare_airline
```

```python
# 2. Average Fare by Destination - Find the average Base Fare for each Destination

average_base_fare_destination = df.groupby("Destination")["Base_Fare"].mean()
average_base_fare_destination
```

```python
# 3. Number of Bookings by Booking Channel - Find how many bookings were made through each: Website, Mobile App, Travel Agent

no_of_bookings_by_booking_channel = df.groupby("Booking_Channel")["Booking_ID"].count()
no_of_bookings_by_booking_channel
```

```python
# 4. Average Customer Rating by Airline - Find the average Customer Rating for each Airline

avg_cust_rating_airline = df.groupby("Airline")["Customer_Rating"].mean()
avg_cust_rating_airline
```

```python
# 5. Airline and Booking Channel Analysis - Find the number of bookings for each Airline and Booking Channel. Use multiple columns with groupby().

no_of_bookings_airline_channel = df.groupby(["Airline", "Booking_Channel"])["Booking_ID"].count()
no_of_bookings_airline_channel
```

```python
# 6. Airline Fare Summary For each Airline, find: Minimum Base Fare, Maximum Base Fare, Average Base Fare. Use agg().

min_max_avg_airline = df.groupby("Airline")["Base_Fare"].agg(["min", "max", "mean"])
min_max_avg_airline
```

```python
# 7. Destination and Cabin Analysis. For each Destination and Cabin_Code, find the average Base Fare.

avg_base_fare_destination_cabin = df.groupby(["Destination", "Cabin_Code"])["Base_Fare"].mean()
avg_base_fare_destination_cabin
```

---
# *Part B – Map*
---

```python
# 8. Convert Cabin Codes
# The Cabin_Code column contains:
# • E 
# • P 
# • B 
# Use a dictionary and map() to create a new column called:
# Cabin_Type
# Mapping:
# • E → Economy 
# • P → Premium Economy 
# • B → Business

cabin_code_mapping = {
    "E" : "Economy",
    "P" : "Premium Economy",
    "B" : "Business"
}

df["Cabin_Type"] = df["Cabin_Code"].map(cabin_code_mapping)
df
```

```python
# 9. Convert Payment Codes
# The Payment_Code column contains:
# • UPI 
# • CC 
# • DC 
# • NB 
# Use map() to create:
# Payment_Method
# Convert the codes into meaningful payment method names.

df["Payment_Method"] = df["Payment_Code"].map({
    "UPI": "Unified Payments Interface",
    "CC": "Credit Card",
    "DC": "Debit Card",
    "NB": "Net Banking"
})

df
```

```python
# 10. Convert Meal Codes
# Use map() on Meal_Code and create:
# Meal_Type
# Convert:
# • VG → Vegetarian 
# • NV → Non-Vegetarian 
# • SF → Special Food

df["Meal_Type"] = df["Meal_Code"].map({
    "VG" : "Vegetarian",
    "NV" : "Non-Vegetarian",
    "SF" : "Special Food"
})

df
```

---
# *Part C – Apply*
---

```python
# 11. Create Baggage Category

def baggage_category(kg):
    if kg <= 15:
        return "Light"
    elif kg <= 25:
        return "Medium"
    else:
        return "Heavy"

df["Baggage_Category"] = df["Baggage_KG"].apply(baggage_category)
df
```

```python
# 12. Create Flight Category

def flight_category(hours):
    if hours < 3:
        return "Short Flight"
    elif hours <= 7:
        return "Medium Flight"
    else:
        return "Long Flight"

df["Flight_Category"] = df["Flight_Duration_Hours"].apply(flight_category)
df
```

```python
# 13. Create Fare Level
df["Fare_Level"] = df["Base_Fare"].apply(
    lambda x: "High Fare" if x >= 25000 else "Normal Fare"
)

df
```

```python
# 14. Create Rating Level

df["Rating Level"] = df["Customer_Rating"].apply(
    lambda x: "Excellent" if x >= 4.5 else "Needs Improvement"
)

df
```

```python
# 15. Calculate Total Booking Cost

df["Total_Booking_Cost"] = df.apply(lambda row: row["Base_Fare"] + row["Meal_Charge"] + row["Service_Fee"], axis=1)
df
```

```python
# 16. Calculate Discount Amount

def discount_amount(row):
    return row["Base_Fare"] * row["Discount_Percent"] / 100

df["Discount_Amount"] = df.apply(discount_amount, axis=1)

df
```

```python
#17. Calculate Final Fare
def final_fare(row):
    return (row["Base_Fare"]
            + row["Baggage_Charge"]
            + row["Meal_Charge"]
            + row["Service_Fee"]
            - row["Discount_Amount"])

df["Final_Fare"] = df.apply(final_fare, axis=1)

df.head(10)
```
