```python
# Create a Pandas Series with monthly electricity usage.
monthly_units = pd.Series(
    [420, 385, 510, 465, 590],
    index=["March", "April", "May", "June", "July"],
    name="Electricity_Units"
)

monthly_units
```

```python
# Access a Series value by label and by position.
print("May usage:", monthly_units["May"])
print("First month's usage:", monthly_units.iloc[0])
```

```python
# Create an attendance Series and calculate its average.
attendance = pd.Series([92, 78, 85, 96, 73], name="Attendance")

print(attendance)
print()
print("Average attendance:", attendance.mean())
```

```python
# Filter orders where the city is Chennai.
chennai_orders = df[df["city"] == "Chennai"]

chennai_orders.head()
```

```python
# Filter orders where net_amount is greater than 100000.
df[df["net_amount"] > 100000].head()
```

```python
# Filter Chennai customers whose net_amount is greater than 50000.
#Find the customers from Chennai whose net_amount is greater than 50,000.

df[
    (df["city"] == "Chennai") &
    (df["net_amount"] > 50000)
].head()
```

```python
# Filter the first five customers from Chennai or Madurai.
#Find the first 5 customers who belong to either Chennai or Madurai.

df[
    (df["city"] == "Chennai") |
    (df["city"] == "Madurai")
].head()
```
